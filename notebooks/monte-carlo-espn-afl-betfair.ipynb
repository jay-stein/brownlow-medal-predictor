## Team ID mapping

```python
player_to_team_merge = afl_player_to_team.merge(betfair_player_to_team)
player_to_team_merge.head()
```

```python
team_crosstab = pd.crosstab(player_to_team_merge['teamId'], player_to_team_merge['TEAM'])
team_crosstab
```

```python
team_max_count = team_crosstab.max(axis=1).to_frame('max_count')
team_max_match = team_crosstab.idxmax(axis=1).to_frame('max_match')
team_max_combo = pd.concat([team_max_count, team_max_match], axis=1)
```

# Line up player with the team

```python
team_id_map = {'ADELAIDE':'Adelaide Crows','BRISBANE':'Brisbane Lions','COLLINGWOOD':'Collingwood','GOLD COAST':'Gold Coast SUNS','CARLTON':'Carlton','NORTH MELBOURNE':'North Melbourne','PORT ADELAIDE':'Port Adelaide',''
'WESTERN BULLDOGS':'Western Bulldogs','HAWTHORN':'Hawthorn','GEELONG':'Geelong Cats','ST. KILDA':'St Kilda','ESSENDON':'Essendon','SYDNEY':'Sydney Swans','FREMANTLE':'Fremantle','GREATER WESTERN SYDNEY':'GWS GIANTS',
'RICHMOND':'Richmond','MELBOURNE':'Melbourne','WEST COAST':'West Coast Eagles'}
```

```python
reversed_team_id_map = {
    'Adelaide Crows': 'ADELAIDE',
    'Brisbane Lions': 'BRISBANE',
    'Collingwood': 'COLLINGWOOD',
    'Gold Coast SUNS': 'GOLD COAST',
    'Carlton': 'CARLTON',
    'North Melbourne': 'NORTH MELBOURNE',
    'Port Adelaide': 'PORT ADELAIDE',
    'Western Bulldogs': 'WESTERN BULLDOGS',
    'Hawthorn': 'HAWTHORN',
    'Geelong Cats': 'GEELONG',
    'St Kilda': 'ST. KILDA',
    'Essendon': 'ESSENDON',
    'Sydney Swans': 'SYDNEY',
    'Fremantle': 'FREMANTLE',
    'GWS GIANTS': 'GREATER WESTERN SYDNEY',
    'Richmond': 'RICHMOND',
    'Melbourne': 'MELBOURNE',
    'West Coast Eagles': 'WEST COAST'
}
```

```python
teams_by_round = teams[teams["round.year"] == 2025]
```

```python
teams_by_round = teams_by_round[['matchId','match.homeTeam.name','match.awayTeam.name','round.name']]
teams_by_round.loc[:,'round.name'] = teams_by_round.loc[:,'round.name'].str.lower().str.replace(' ','_')
teams_by_round['round.name'] = teams_by_round['round.name'].replace('opening_round', 'round_0')
```

```python
teams_by_round['home_team'] = teams_by_round['match.homeTeam.name'].map(reversed_team_id_map)
teams_by_round['away_team'] = teams_by_round['match.awayTeam.name'].map(reversed_team_id_map)
```

```python
teams_by_round.sample(n=3)
```

# Line up Predictions

## [1] ESPN

```python
espn_prediction = (espn_long
 .merge(afl[['PLAYER','teamId']])
 .merge(team_max_combo.reset_index())
 .drop(columns=['teamId','max_count'])
 .assign(VOTES=lambda df: pd.to_numeric(df['VOTES'].replace('-', '0'), errors='coerce'))
)
```

```python
espn_prediction_wide = pd.concat([espn_prediction.merge(teams_by_round, left_on=['max_match','ROUND_ID'], right_on=['home_team','round.name']),
           espn_prediction.merge(teams_by_round, left_on=['max_match','ROUND_ID'], right_on=['away_team','round.name'])])
```

```python
# SET ROUND 24 to 0 - to avoid double count
espn_prediction_wide.loc[(espn_prediction_wide["matchId"] == 'CD_M20250142401') & (espn_prediction_wide['max_match'] == 'GOLD COAST'), 'VOTES'] = 0
```
