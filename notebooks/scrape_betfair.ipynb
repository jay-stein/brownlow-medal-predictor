```python
import urllib.request, json
import pandas as pd
```

```python
def get_json(url):
    with urllib.request.urlopen(url) as resp:
        return json.load(resp)
```

```python
brownlow_data_betfair = get_json("https://betfair-data-supplier-prod.herokuapp.com/api/brownlow?year=2025&widget=brownlow")
```

```python
selection_df = pd.json_normalize(brownlow_data_betfair)
```

```python
votes_df = pd.json_normalize(brownlow_data_betfair, record_path='votes', meta=['selection_id','name'])
```
