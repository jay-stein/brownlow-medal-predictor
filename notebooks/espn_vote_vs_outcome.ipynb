# Read in ESPN predictions

```python
espn_file = r"C:\Users\mrjay\python\jupyter_notebooks\Projects\brownlow\data\espn_historical_predictions.xlsx"
espn_2024 = pd.read_excel(espn_file, sheet_name="espn_2024")
espn_2023 = pd.read_excel(espn_file, sheet_name="espn_2023")
espn_2022 = pd.read_excel(espn_file, sheet_name="espn_2022")
espn_2021 = pd.read_excel(espn_file, sheet_name="espn_2021")
```

```python
espn_2024 = espn_2024.assign(SEASON = 2024)
espn_2023 = espn_2023.assign(SEASON = 2023)
espn_2022 = espn_2022.assign(SEASON = 2022)
espn_2021 = espn_2021.assign(SEASON = 2021)
```

```python
espn_historical = [espn_2021, espn_2022, espn_2023, espn_2024]
```

```python
def norm_cols(df):
    df.columns = df.columns.str.upper() # capitalize
    df.columns = df.columns.str.replace(' ','_')
    df.columns = df.columns.str.replace('.','_')
    return df
```

```python
# normalize columns
espn_historical = [norm_cols(f) for f in espn_historical]
```

```python
# combine list of DF to one single DF
espn_historical_df = pd.concat(espn_historical)
```

## Read in Brownlow Results

```python
brownlow_file = r"c:\Users\mrjay\python\jupyter_notebooks\Projects\brownlow\data\brownlow_stats_2018_2024_fitzroy.csv"
```

```python
brownlow = pd.read_csv(brownlow_file, 
                       usecols=['Brownlow.Votes','First.name','Surname', 'Season','Round','Playing.for','Away.team'])
```

```python
# normalize columns
brownlow = norm_cols(brownlow)

```

```python
# filter as we only care about votes
brownlow = brownlow[brownlow['BROWNLOW_VOTES'] > 0]
```

```python
# convert to integer
brownlow['ROUND'] = brownlow['ROUND'].astype('int')
```

```python
# add player col to join on
brownlow = brownlow.assign(PLAYER = lambda x: (x['FIRST_NAME'] + " " + x['SURNAME']).str.upper())
```
